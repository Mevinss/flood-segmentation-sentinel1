# Модельді жақсарту: шекара қателері, суы аз фрагменттер, бірнеше seed

Бұл ноутбук курстық жұмыстағы үш әлсіз жерді түзетуге арналған:

| Әлсіз жер | Түзету |
|---|---|
| Су шекарасындағы пиксельдердің ~1/3 бөлігі өткізіліп алынады | Шекараға салмақ беретін BCE + Lovász шығыны; 512×512 толық фрагментпен болжау; TTA (8 бұрылыс/шағылыс) |
| Суы аз фрагменттерде нәтиже нашар | Per-image Lovász шығыны: әр фрагмент шығынға тең үлес қосады |
| Бір ғана seed | Әр модель 3 seed-пен (42, 7, 2024) оқытылады, нәтиже «орташа ± ауытқу» түрінде беріледі |

**Ереже:** шек, болжау режимі және архитектура тек **валидация** бойынша таңдалады. Тест пен Боливия соңғы кестеде ғана қаралады.

**Уақыт (Tesla T4):** бір оқыту ≈ 25–35 мин. 1-кезең: 4 оқыту (≈2 сағ), 2-кезең: бағалау (≈30–40 мин), 3-кезең: 3 оқыту (≈1,5 сағ).
Барлығы Google Drive-қа сақталады. Сеанс үзілсе, ұяшықты қайта іске қоссаңыз, оқыту соңғы дәуірден жалғасады.
Уақыт аз болса, `SEEDS = [42, 7]` қойыңыз (2 seed).

**Runtime → Change runtime type → T4 GPU** таңдаңыз, содан кейін ұяшықтарды ретімен орындаңыз.

$L = \mathrm{BCE}_w + \mathrm{Lovász}_{\text{per-image}}$

* $\mathrm{BCE}_w$: су/құрлық шекарасынан 2 пиксель қашықтықтағы пиксельдердің салмағы 3 есе (1 + 2).
* Lovász hinge (Berman et al., 2018) IoU-ды тікелей оңтайландырады. Ол әр фрагмент үшін бөлек есептеледі, сондықтан суы аз фрагменттер ірі су айдындарының көлеңкесінде қалмайды.

Архитектура мен басқа параметрлер өзгермейді, тек шығын функциясы ауысады.

```python
BL = {'loss': 'boundary_lovasz', 'boundary_weight': 2.0, 'boundary_radius': 2}
for seed in SEEDS:
    start = time.time()
    ck = train_v2(df, norm, run_dir(BEST_ARCH, seed, '_bl'), {'model': BEST_ARCH, 'pretrained': True, 'seed': seed, **BL})
    print(f'{BEST_ARCH}+BL seed {seed}: {(time.time() - start) / 60:.1f} мин')

rows = []
for seed in SEEDS:
    ck = run_dir(BEST_ARCH, seed, '_bl') / 'best.pt'
    for name in dict.fromkeys(['patch', BEST_INFERENCE]):
        mode, tta = VARIANTS[name]
        s = evaluate_v2(df, ck, OUT / 'eval' / f'{BEST_ARCH}_bl_seed{seed}' / name, mode=mode, tta=tta)
        rows += flatten(BEST_ARCH + '+BL', seed, name, s)
stage3 = pd.DataFrame(rows); stage3.to_csv(OUT / 'stage3_all.csv', index=False)
from flood.evaluate import learning_curves
learning_curves(run_dir(BEST_ARCH, 42, '_bl'))
display(Image(filename=str(run_dir(BEST_ARCH, 42, '_bl') / 'learning_curves.png')))
```

## 6. Қорытынды кесте: валидация, тест және Боливия (орташа ± стандартты ауытқу)

Соңғы модель **валидация** нәтижесі бойынша таңдалады. Тест пен Боливия тек есепке жазу үшін көрсетіледі.

### Суы аз фрагменттер және сол seed бойынша жұптық салыстыру

```python
bins = [c for c in allres.columns if c.startswith('iou_water_')]
low = allres[allres.inference == BEST_INFERENCE].groupby(['model', 'split'])[bins].mean().round(4)
display(low)

base = allres[(allres.inference == 'patch') & (allres.model == 'resnet34')].set_index(['seed', 'split']).iou
for m in allres.model.unique():
    if m == 'resnet34':
        continue
    cur = allres[(allres.inference == BEST_INFERENCE) & (allres.model == m)].set_index(['seed', 'split']).iou
    d = (cur - base).dropna().unstack('split') * 100
    print(f'\n{m} ({BEST_INFERENCE}) − ResNet-34 (patch), пайыздық тармақ, әр seed:')
    display(d.round(2))
    print('орташа:', d.mean().round(2).to_dict(), ' барлық seed-те оң ба:', (d > 0).all().to_dict())

tile = 'Bolivia_242570'
rows = []
for d in sorted((OUT / 'eval').glob('*/' + BEST_INFERENCE)) + sorted((OUT / 'eval').glob('*/patch')):
    t = pd.read_csv(d / 'per_tile.csv'); r = t[t.chip_id == tile]
    if len(r): rows.append({'run': d.parent.name, 'inference': d.name, **r.iloc[0][['iou', 'precision', 'recall', 'fp', 'fn']].to_dict()})
display(pd.DataFrame(rows).sort_values(['inference', 'run']) if rows else f'{tile} табылмады')
```

### Есепке арналған график

```python
import matplotlib.pyplot as plt, numpy as np
plot = keep[keep.split.isin(['test', 'bolivia'])]
plot = plot[(plot.inference == BEST_INFERENCE) | ((plot.model == 'resnet34') & (plot.inference == 'patch'))]
g = plot.groupby(['model', 'inference', 'split']).iou.agg(['mean', 'std']).reset_index()
g['label'] = g.model.str.replace('attention_resnet34', 'Attention U-Net').str.replace('resnet34', 'ResNet-34 U-Net') + ' (' + g.inference + ')'
fig, ax = plt.subplots(figsize=(10, 4.5))
labels = list(dict.fromkeys(g.label)); xs = np.arange(len(labels)); w = .38
for i, (split, name) in enumerate([('test', 'Негізгі тест'), ('bolivia', 'Боливия')]):
    s = g[g.split == split].set_index('label').reindex(labels)
    ax.bar(xs + (i - .5) * w, s['mean'] * 100, w, yerr=s['std'] * 100, capsize=4, label=name)
ax.set_xticks(xs); ax.set_xticklabels(labels, rotation=15, ha='right', fontsize=8)
ax.set_ylabel('IoU, % (орташа ± ст. ауытқу)'); ax.set_ylim(50, 80); ax.legend(); ax.grid(axis='y', alpha=.3)
ax.set_title(f'Су сегментациясы, {len(SEEDS)} seed')
fig.tight_layout(); fig.savefig(OUT / 'iou_mean_std.png', dpi=300); plt.show()
print('Нәтижелер сақталды:', OUT)
```
